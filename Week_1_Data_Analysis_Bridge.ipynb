**Part A -- Translation (Conceptual)**

For the R code below:

```
x <- c(1, 3, 5, 7)
y <- ifelse(x > 4, x^2, x + 1)
mean(y)
```




Predict the output:

"x" is a numeric vector containing numbers 1, 3, 5, 7. If numbers in x are greater than 4, they are squared; if they are less than 4, they are increased by 1. Performing these operation should return y = (2, 4, 25, 49) The mean of "y" is 20.

Translation & Computation in Python:

````python

import numpy as np
x = np.array([1, 3, 5, 7])
y = np.where(x > 4, x**2, x + 1)
np.mean(y)
````

**Part B -- Lists vs NumPy Arrays**

Create a Python list of numbers 1--5\
Attempt to multiply the list by 2\
Convert the list to a NumPy array and repeat the multiplication
Answer the following questions:

What happened in each case?\
Why does NumPy behave differently?


````python
#Creating a list without converting to NumPy array
a = [1, 2, 3, 4, 5]
b = a * 2
print(b)
````

The list duplicated itself instead of multiplying every number by 2.

````python
#Converting to NumPy array
a = np.array([1, 2, 3, 4, 5])
b = a * 2
print(b)
````

After converting the list to a NumPy array, each number was multiplied by 2. NumPy behaves differently because it turns a list into a  vector, so mathematical operations are automatically applied to each element.
